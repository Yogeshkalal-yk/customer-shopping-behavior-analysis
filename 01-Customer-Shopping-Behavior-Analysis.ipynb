```python
# 1. df.isna()  Checks every value in the DataFrame. Returns True if a value is missing.Returns False if a value is present.
# 2. .sum() Counts the number of True values in each column. In Python, True is treated as 1 and False as 0.

df.isnull().sum()  # is used to check how many missing (empty or NaN) values are present in each column of a dataset.
```

```python
df['Review Rating'].median()
```

```python
df.groupby('Category')['Review Rating'].median()
```

```python
# Review ratings are bounded (usually 1 to 5) and often skewed, since people tend to leave mostly high ratings. The median stays at a realistic value, while the mean can land on something odd like 3.74.
# Better fill (median by category)
# Ratings often differ by product type, so filling with the median of the same Category is more accurate:

df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isna().sum()
```

```python
# make to all calummn in sneck_case(lower case and _ in space) bez realble and and fetch data in python or sql withount error 

df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df = df.rename(columns = {'purchase_amount_(usd)':'purchase_amount'})
```

```python
list(df.columns)

```

```python
# feature engineering
# create the age gorup column for spit the age gorup between 4 equal parts for undersating customer behevor wich age gorup purche the frequly 
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q = 4, labels = labels) #qcut is divide data into the n equal parts
```

```python
df[['age', 'age_group']].head(10)
```

```python
# create column purchase_frequency_days 
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

#The map() function in Python is used to apply a function to every item in an iterable (like a list or tuple).
# Iterable means an object that you can go through one item at a time using a loop.
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```
